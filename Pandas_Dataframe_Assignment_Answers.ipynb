Q4. Indexing using .loc

Using .loc, perform the following:

1. Display the row with index 10.

2. Display the Sales value of the row with index 20

```python
#A4.
# 1. Display the row with index 10
display(df.loc[10])

# 2. Display the Sales value of the row with index 20
sales_val_20 = df.loc[20, 'Sales']
print("Sales value at index 20:", sales_val_20)
```

Q5. Indexing using .iloc

Using .iloc, display:

1. The first 6 rows and first 4 columns.

2. The Sales value from the 3rd row.

```python
#A5.
# 1. The first 6 rows and first 4 columns
display(df.iloc[:6, :4])

# 2. The Sales value from the 3rd row (Index 2 represents the 3rd row)
sales_col_index = df.columns.get_loc('Sales')
sales_val_3rd_row = df.iloc[2, sales_col_index]

print("Sales value from the 3rd row:", sales_val_3rd_row)
```

Q7. Filtering Data

Filter the dataset to display:

1. Rows where Region = "East".

2. Rows where Sales > 1500.

```python
#A7.
# 1. Rows where Region = "East"
east_region = df[df['Region'] == 'East']
display(east_region)

# 2. Rows where Sales > 1500
high_sales = df[df['Sales'] > 1500]
display(high_sales)
```

Q8. Multiple Condition Filtering

Display rows where:

* Region is West

* Sales is greater than 1200

```python
#A8.
# Rows where Region is West AND Sales is greater than 1200
west_high_sales = df[(df['Region'] == 'West') & (df['Sales'] > 1200)]
display(west_high_sales)
```

Q10. Aggregation Analysis

Using the Sales column, calculate:

1. Total sales

2. Average sales

3. Maximum sales

3. Minimum sales

```python
#A10.
# 1. Total sales
total_sales = df['Sales'].sum()
print("Total Sales:", total_sales)

# 2. Average sales
average_sales = df['Sales'].mean()
print("Average Sales:", average_sales)

# 3. Maximum sales
max_sales = df['Sales'].max()
print("Maximum Sales:", max_sales)

# 4. Minimum sales
min_sales = df['Sales'].min()
print("Minimum Sales:", min_sales)
```
